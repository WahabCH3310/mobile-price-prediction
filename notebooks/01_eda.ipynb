## 1. Load the data
`build_dataset()` loads the raw Kaggle CSV and attaches the demonstration proxy `price_usd` for the regression sub-task.

```
from src.data.make_dataset import build_dataset
from src.features.build_features import add_engineered_features

df = add_engineered_features(build_dataset(save=False))
print(df.shape)
df.head()
```

## 2. Class balance
The classification target `price_range` is perfectly balanced (500 phones per tier).

```
df['price_range'].value_counts().sort_index()
```

## 3. Price distribution & spec relationships

```
fig, ax = plt.subplots(1, 2, figsize=(13,4))
sns.histplot(df['price_usd'], bins=40, kde=True, ax=ax[0])
ax[0].set_title('price_usd distribution (proxy)')
sns.scatterplot(data=df, x='ram', y='price_usd', hue='price_range',
                palette='viridis', alpha=0.6, ax=ax[1])
ax[1].set_title('RAM vs price'); plt.tight_layout()
```

## 4. Correlation heatmap

```
plt.figure(figsize=(13,10))
sns.heatmap(df.select_dtypes('number').corr(), cmap='coolwarm',
            center=0, square=True, cbar_kws={'shrink':0.7})
plt.title('Correlation heatmap'); plt.show()
```

**Key takeaway:** `ram` has by far the strongest correlation with `price_range` — a linear relationship that lets even Logistic Regression classify tiers very accurately.
